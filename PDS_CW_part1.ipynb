## Q1.1: Reading in rows of a CSV file. [10 marks]

Write a function `csv_rows(filename, from_row=None, to_row=None)`
to read in the header and specific rows of a CSV file.
The input arguments of the function should be a string specifying a filename,
a starting row number
`from_row` and
a final row number `to_row`.
The row numbers should be either integers or None.
The starting row number can be 1 or higher or None.
Row 1 refers to the first row after the header.
If the starting row number is None, then the default value of 1 should be used.
The ending row number can be 1 or higher or None.
The ending row number refers to the last row to be included.
If the ending row number is None, then all rows starting at the starting row should be read in.
The function should return two outputs:
a list with the column names and list of rows, where each row is a list of values (strings or numbers).

You can use the file `part1.csv` for your tests.

### Explanation for Q1.1

To read the CSV file without external libraries, we use the built-in `open()` function. 
1. We read all lines using `readlines()`.
2. The first line is treated as the `header`, which we split by commas after removing leading/trailing whitespace.
3. We slice the list of lines from `from_row` to `to_row`. Note that `from_row=1` corresponds to index 1 in the lines list (the first line after the header).
4. For each row, we attempt to convert values into `int` or `float`. If a value cannot be converted (like '-' or text), we catch the `ValueError` and keep it as a string.

```python
filename= 'part1.csv'
def csv_rows(filename, from_row=None, to_row=None):
    """
    Reads a CSV file and returns the header and a list of rows.
    Rows are converted to numbers where possible.
    """
    if from_row is None:
        from_row = 1
    
    try:
        with open(filename, 'r') as f:
            lines = f.readlines()
    except FileNotFoundError:
        return None, None

    if not lines:
        return None, None

    header = lines[0].strip().split(',')
    
    start_idx = from_row
    if to_row is None:
        end_idx = len(lines)
    else:
        end_idx = to_row + 1
        
    data_lines = lines[start_idx:end_idx]
    
    rows = []
    for line in data_lines:
        raw_values = line.strip().split(',')
        processed_row = []
        for val in raw_values:
            try:
                if '.' in val:
                    processed_row.append(float(val))
                else:
                    processed_row.append(int(val))
            except ValueError:
                processed_row.append(val)
        rows.append(processed_row)
            
    return header, rows
header, rows = csv_rows(filename, from_row=1, to_row=5)
print("Header:", header)
print("Rows:", rows[0])
```

## Q1.2: Reading in Columns of a CSV file. [10 marks]

Write a function `csv_cols(filename, from_row=None, to_row=None, columns=None)`
that takes as inputs a filename, a starting row, a final row and a list of column names.
The same conventions as for Q1.1 should be used for the starting and final row input parameters.
The column name parameter can either be a list of column names or None, which indicates all columns.
The function should return a dictionary with the selected column names as
keys and a list of values (strings or numbers) for the corresponding column.
Your function should use the function from Q1.1.

You can use the file `part1.csv` for your tests.



### Explanation for Q1.2

The function `csv_cols` transforms the row-based data into column-based data.
1. It calls `csv_rows` to get the header and the data rows.
2. It initializes a dictionary where each key is a column name from the header, and each value is an empty list.
3. It iterates through the rows and appends each value to the corresponding list in the dictionary using the index.
4. Finally, if a specific list of `columns` is provided, it filters the dictionary to only include those keys. (Youtube, 2020)

```python
def csv_cols(filename, from_row=None, to_row=None, columns=None):
    """
    Reads a CSV file and returns a dictionary where keys are column names 
    and values are lists of column data.
    """
    header, rows = csv_rows(filename, from_row, to_row)
    if not header or not rows:
        return {}
    
    # Create dictionary mapping header names to lists of values
    col_dict = {name: [] for name in header}
    
    for row in rows:
        for i, val in enumerate(row):
            if i < len(header):
                col_dict[header[i]].append(val)
                
    # Filter if specific columns are requested
    if columns is not None:
        filtered_dict = {name: col_dict[name] for name in columns if name in col_dict}
        return filtered_dict
    
    return col_dict

cols = csv_cols('part1.csv', from_row=1, to_row=3, columns=['10-14 years old', '15-19 years old'])
print("Selected Columns Data:", cols)
```

## Q1.3: Calculating the minimum, maximum and median of a list of values [10 marks]

Write a function `min_max_median(p)` to calculate
the minimum, maximum and median of a list of values.
The function should return the minimum, maximum and median.



### Explanation for Q1.3

The function `min_max_median` calculates basic statistics for a list, it describes how data is being distributed.
1. **Filtering**: Since the dataset may contain missing values (like '-'), we first filter the input list to keep only integers and floats.
2. **Sorting**: We sort the numeric list. The minimum is then at index 0 and the maximum is at the last index (`-1`).
3. **Median**: 
   - If the length is odd, the median is the middle element.
   - If the length is even, the median is the average of the two middle elements.

```python
def min_max_median(p):
    """
    This calculates the minimum, maximum, and median of a list.
    Non-numeric values are ignored.
    """
    # Filter for numbers only
    numeric_values = [x for x in p if isinstance(x, (int, float))]
    
    if not numeric_values:
        return None, None, None
    
    sorted_list = sorted(numeric_values)
    n = len(sorted_list)
    
    minimum = sorted_list[0]
    maximum = sorted_list[-1]
    
    # Formula for finding Median 
    if n % 2 == 1:
        median = sorted_list[n // 2]
    else:
        m1 = sorted_list[n // 2 - 1]
        m2 = sorted_list[n // 2]
        median = (m1 + m2) / 2
        
    return minimum, maximum, median

test_data = [17, 6.5, 10, '-', 7, 1]
min, max, med = min_max_median(test_data)
print(f"Min: {min}")
print(f"Max: {max}")
print(f"Median: {med}")
```

## Q1.4: Calculating the Minimum, Maximum and Median of Rows and Columns of a Table [10 marks]

Write a function `table_min_max_median(data, columns=None)` that takes as inputs
a dictionary with column names as keys and a list of values for each key.
The `columns` parameter can be either a list of column names or None, which
means all column names should be used.
The function should return a list of (minimum, maximum, median) tuples for each row,
and a list of (minimum, maximum, median) tuples for the selected columns.
Your function must use the function from Q1.3.

You can use the file `part1.csv` and the function from Q1.2 for your tests.

### Explanation for Q1.4

This function applies `min_max_median` across both rows and columns.
1. **Column Stats**: For the column stats, iteration was carried out through the requested columns in the dictionary and `min_max_median` was applied to each column list.
2. **Row Stats**: As for the row stats, iteration was through the indices of the lists. For each index `i`, we gather the values from all selected columns at that index to form a virtual "row" and then its stats was calculated. (Youtube, 2021)

```python
def table_min_max_median(data, columns=None):
    """
    Calculations for stats for each row and each column that is selected.
    """
    if columns is None:
        selected_columns = list(data.keys())
    else:
        selected_columns = [c for c in columns if c in data]

    # Calculation forr column stats
    col_stats = []
    for col_name in selected_columns:
        col_stats.append(min_max_median(data[col_name]))
    
    # Calculation for row stats
    if not selected_columns:
        return [], []
        
    num_rows = len(data[selected_columns[0]])
    row_stats = []
    for i in range(num_rows):
        row_values = [data[col_name][i] for col_name in selected_columns]
        row_stats.append(min_max_median(row_values))
        
    return row_stats, col_stats

data_dict = csv_cols('part1.csv', from_row=1, to_row=10)
row_s, col_s = table_min_max_median(data_dict, columns=['10-14 years old', '15-19 years old'])
print("First Row Stats:", row_s[0])
print("First Column Stats:", col_s[0])
```

## Q1.5: Develop a Function to Print a Custom Table [10 marks]

Write a function `print_min_max_median(row_stats, col_stats, ...)`
that prints the results in a clear and attractive way.
The input parameters should be a list of (min, max, median) tuples for the rows
and a list of (min, max, median) tuples for the columns
(as generated by Q1.4).
You can add extra parameters for further customisation.

You can use the file `part1.csv` and the functions from Q1.2 and Q1.4 for your tests.

### Explanation for Q1.5

To display the results clearly, we use string formatting.
1. We print a header for the table with alignment (`<` for left, `>` for right).
2. We use f-strings with precision formatting (e.g., `:.2f`) to ensure the numbers look consistent.
3. We include separators to make the output visually organized.

```python
def print_min_max_median(row_stats, col_stats, column_names=None):
    """
    Prints a summary table of the calculated statistics.
    """
    print("="*60)
    print(f"{'Item':<25} | {'Min':>8} | {'Max':>8} | {'Median':>8}")
    print("-"*60)
    
    # Column Statistics
    if column_names:
        for name, stats in zip(column_names, col_stats):
            min, max, med = stats
            print(f"Col: {name[:20]:<20} | {min:>8.2f} | {max:>8.2f} | {med:>8.2f}")
    
    print("-"*60)
    
    # First few Row Statistics
    for i, stats in enumerate(row_stats[:5]):
        min, max, med = stats
        print(f"Row {i+1:<20} | {min:>8.2f} | {max:>8.2f} | {med:>8.2f}")
    
    if len(row_stats) > 5:
        print(f"... and {len(row_stats)-5} more rows.")
    print("="*60)

cols_to_test = ['10-14 years old', '25-29 years old', '70+ years old']
final_data = csv_cols('part1.csv', from_row=1, to_row=20, columns=cols_to_test)
rs, cs = table_min_max_median(final_data, columns=cols_to_test)
print_min_max_median(rs, cs, column_names=cols_to_test)
```
